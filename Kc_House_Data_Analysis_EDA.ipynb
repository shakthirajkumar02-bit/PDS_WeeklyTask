```python
df["date"] = pd.to_datetime(
    df["date"],
    format="%Y%m%dT%H%M%S",
    errors="coerce"
)

print("Date conversion completed!")
```

```python
df["sale_year"] = df["date"].dt.year

df[["date", "sale_year"]].head()
```

```python
df["property_age"] = df["sale_year"] - df["yr_built"]

df[["yr_built", "sale_year", "property_age"]].head()
```

```python
df["is_renovated"] = df["yr_renovated"] > 0

df[["yr_renovated", "is_renovated"]].head(10)
```

```python
df["price_per_sqft_living"] = (
    df["price"] / df["sqft_living"]
)

df[["price", "sqft_living", "price_per_sqft_living"]].head()
```

```python
df["price_per_sqft_lot"] = (
    df["price"] / df["sqft_lot"]
)

df[["price", "sqft_lot", "price_per_sqft_lot"]].head()
```

```python
zero_bedrooms = df[df["bedrooms"] == 0]

print("Houses with zero bedrooms:", len(zero_bedrooms))
```

```python
invalid_rooms = df[
    (df["bedrooms"] == 0) |
    (df["bathrooms"] == 0)
]

print("Total records to investigate:", len(invalid_rooms))

invalid_rooms.head()
```

```python
clean_df = df.copy()

print("Original rows:", len(df))
print("Clean dataset created!")
```

```python
clean_df = clean_df.dropna(
    subset=["price", "sqft_living", "sqft_lot"]
)

print("Rows after cleaning:", len(clean_df))
```

```python
clean_df["zero_bedroom_flag"] = (
    clean_df["bedrooms"] == 0
)

print(clean_df["zero_bedroom_flag"].value_counts())
```

```python
Q1 = clean_df["price"].quantile(0.25)
Q3 = clean_df["price"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
price_outliers = clean_df[
    (clean_df["price"] < lower_limit) |
    (clean_df["price"] > upper_limit)
]

print("Number of Price Outliers:", len(price_outliers))
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    clean_df["sqft_living"],
    clean_df["price"],
    alpha=0.4
)

plt.title("Living Area vs House Price")
plt.xlabel("Living Area (sqft)")
plt.ylabel("House Price")

plt.show()
```

```python
x = clean_df["sqft_living"]
y = clean_df["price"]

slope, intercept = np.polyfit(x, y, 1)

predicted_price = slope * x + intercept

print("Slope:", slope)
print("Intercept:", intercept)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(x, y, alpha=0.3, label="Actual Prices")

sorted_index = x.sort_values().index

plt.plot(
    x.loc[sorted_index],
    predicted_price.loc[sorted_index],
    color="red",
    label="Regression Line"
)

plt.title("House Price Prediction")
plt.xlabel("Living Area (sqft)")
plt.ylabel("House Price (USD)")
plt.legend()
plt.show()
```

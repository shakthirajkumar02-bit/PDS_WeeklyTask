```python
text_columns = ["Category", "Sub-Category", "Segment"]

for column in text_columns:
    df[column] = (
        df[column]
        .astype(str)
        .str.strip()
        .str.title()
    )
```

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Category", 
    y="Sales", 
    estimator="sum"
)

plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df, 
    x="Category",
    y="Profit",
    estimator="sum"
)

plt.title("Category-wise Total Profit")
plt.xlabel("category")
plt.ylabel("Total Profit")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df, 
    y="Sales"
)

plt.title("Sales Distribution")
plt.ylabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    y="Profit"
)

plt.title("Profit Distribution")
plt.ylabel("Profit")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df, 
    x="Discount", 
    y="Profit"
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

```python
df["Profit Margin"] = (df["Profit"] / df["Sales"]) * 100

df[["Sales", "Profit", "Profit Margin"]].head()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df, 
    x="Discount", 
    y="Profit Margin"
)

plt.title("Discount vs Profit Margin")
plt.xlabel("discount")
plt.ylabel("profit Margin (%)")

plt.show()
```

```python
numerical_df = df.select_dtypes(include=np.number)

correlation_matrix = numerical_df.corr()

correlation_matrix
```

```python
plt.figure(figsize=(18, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Numerical attributes")

plt.show()
    
```
